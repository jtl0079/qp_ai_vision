```python
from rsw_ai.backend.import_sshikamaru_car_object_detection_dataset import (
    import_sshikamaru_car_object_detection_dataset,
)
from rsw_ai.mapper.VisionYoloDatasetMapper import VisionYoloDatasetMapper
from rsw_ai.model.YoloExporter import YoloExporter

# import
vision_dataset = import_sshikamaru_car_object_detection_dataset(
    "../outputs/datasets/downloads/kaggle/sshikamaru_car_object_detection/data/"
)
# parse
yolo_dataset = VisionYoloDatasetMapper.from_VisionDetectionDataset(
    dataset=vision_dataset, image_width=676, image_height=380
)

# export
yolo_exporter = YoloExporter()

yolo_exporter.export_dataset("../outputs/datasets/yolo/sshikamaru_car_object_detection/", yolo_dataset)
```

```python
from pathlib import Path

from ultralytics import YOLO

# ====================================
# Paths
# ====================================

model_path = Path("../outputs/models/yolo/yolo11n.pt").resolve()

dataset_yaml_path = Path(
    "../outputs/datasets/yolo/sshikamaru_car_object_detection/dataset.yaml"
).resolve()

experiment_path = Path(
    "../outputs/experiments"
).resolve()

# ====================================
# Load pretrained model
# ====================================

model = YOLO(
    model_path,
)

# ====================================
# Train
# ====================================

results = model.train(
    data=str(dataset_yaml_path),

    project=str(experiment_path),

    name="sshikamaru_car_object_detection",

    epochs=50,
    imgsz=320,
    batch=2,
    device="cpu",
    workers=0,
)

print(results)
```

```python
from pathlib import Path

from ultralytics import YOLO


model_path = Path("../outputs/experiments/sshikamaru_car_object_detection/weights/best.pt").resolve()

image_path = Path( "../outputs/datasets/downloads/kaggle/sshikamaru_car_object_detection/data/testing_images/vid_5_400.jpg").resolve()

# ====================================
# Load pretrained model
# ====================================

model = YOLO(
    model_path,
)

print(model.ckpt_path)

results = model.predict(

    source=image_path,
    imgsz=640,
    conf=0.25,
    save=True,
    show=True
)

for box in results[0].boxes:
    print("类别:", int(box.cls))
    print("名称:", results[0].names[int(box.cls)])
    print("置信度:", float(box.conf))
    print("坐标:", box.xyxy.tolist())
    print("-" * 30)
```
