```python
from kaggle.api.kaggle_api_extended import KaggleApi
from rsw_ai.enum.DatasetRepositoryLayout import DatasetRepositoryLayout

KAGGLE_DOWNLOADED_PATH = "outputs/datasets/downloads/kaggle/"

DIR_DOWNLOADED_SSHIKAMARU_CAR = (
    "../"
    + KAGGLE_DOWNLOADED_PATH
    + DatasetRepositoryLayout.SSHIKAMARU_CAR_OBJECT_DETECTION.label
    + "/"
)

api = KaggleApi()
api.authenticate()

api.dataset_download_files(
    "sshikamaru/car-object-detection",
    path=DIR_DOWNLOADED_SSHIKAMARU_CAR,
    unzip=True,
)

print(DIR_DOWNLOADED_SSHIKAMARU_CAR)
```

```python
from kaggle.api.kaggle_api_extended import KaggleApi
from rsw_ai.enum.DatasetRepositoryLayout import DatasetRepositoryLayout


# ====================================
# Initiate Variable
# ====================================

KAGGLE_DOWNLOADED_PATH = "outputs/datasets/downloads/kaggle/"

dataset_repository_layout = (
    DatasetRepositoryLayout.NADINPETHIYAGODA_VEHICLE_DATASET_FOR_YOLO
)

DIR_DOWNLOADED_SSHIKAMARU_CAR = (
    "../" + KAGGLE_DOWNLOADED_PATH + dataset_repository_layout.label + "/"
)

api = KaggleApi()
api.authenticate()

api.dataset_download_files(
    dataset_repository_layout.dataset_name,
    path=DIR_DOWNLOADED_SSHIKAMARU_CAR,
    unzip=True,
)

print(DIR_DOWNLOADED_SSHIKAMARU_CAR)
```
